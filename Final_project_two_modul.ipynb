```python
query = '''
WITH set_config_precode AS (
  SELECT set_config('synchronize_seqscans', 'off', true)
),

purchase_with_prev AS (
  SELECT 
    *,
    created_dt_msk::date - LAG(created_dt_msk::date ) 
        OVER(PARTITION BY user_id ORDER BY created_ts_msk)
        AS days_since_prev
FROM afisha.purchases
WHERE
    
    device_type_canonical IN ('mobile','desktop')
)
SELECT 
  pwp.user_id,
  pwp.device_type_canonical,
  pwp.order_id,
  pwp.created_dt_msk AS order_dt,
  pwp.created_ts_msk AS order_ts,
  pwp.currency_code,
  pwp.revenue,
  pwp.tickets_count,
  pwp.days_since_prev,
  pwp.event_id,
  e.event_name_code AS event_name,
  e.event_type_main,
  pwp.service_name,
  r.region_name,
  c.city_name
FROM purchase_with_prev pwp
INNER JOIN afisha.events e ON pwp.event_id = e.event_id
INNER JOIN afisha.city c ON e.city_id = c.city_id
INNER JOIN afisha.regions r ON c.region_id = r.region_id
WHERE e.event_type_main != 'фильм'
ORDER BY pwp.user_id ASC, order_ts ASC;
'''
```

Для этого используйте датасет с информацией о курсе казахстанского тенге по отношению к российскому рублю за 2024 год — `final_tickets_tenge_df.csv`. Его можно загрузить по пути `https://code.s3.yandex.net/datasets/final_tickets_tenge_df.csv')`

Значения в рублях представлено для 100 тенге.

Результаты преобразования сохраните в новый столбец `revenue_rub`.

```python
ruble_exchange_rate = pd.read_csv('https://code.s3.yandex.net/datasets/final_tickets_tenge_df.csv')
ruble_exchange_rate['data'] = pd.to_datetime(ruble_exchange_rate['data'])
```

```python
external_dict = ruble_exchange_rate.set_index('data').to_dict('index')

def conversion_to_rubles(row):
    if row['currency_code'] == 'kzt':
        date = row['order_dt']
        
        if date in external_dict:
            
            return round((external_dict[date]['curs'] / external_dict[date]['nominal']) * row['revenue'], 2)
        else:
            return float('nan')
    elif row['currency_code'] == 'rub':
        return row['revenue']
df['revenue_rub'] = df.apply(conversion_to_rubles, axis=1)
df.head(5)
```

- Подсчитайте общее количество пользователей в каждом сегменте и их долю в разрезе каждого признака. Сегмент — это группа пользователей, объединённых определённым признаком, то есть объединённые принадлежностью к категории. Например, все клиенты, сделавшие первый заказ с мобильного телефона, — это сегмент.
- Ответьте на вопрос: равномерно ли распределены пользователи по сегментам или есть выраженные «точки входа» — сегменты с наибольшим числом пользователей?

---


```python
def analyze_user_distribution(df_users, group_col, target_col='user_id'):
    # Группировка и подсчет
    distribution = df_users.groupby(group_col)[target_col].count().reset_index()
    distribution.columns = [group_col, 'user_count']
    
    # Добавляем долю
    distribution['share'] = distribution['user_count'] / distribution['user_count'].sum()
    distribution['share_pct'] = distribution['share'] * 100
    distribution = distribution.sort_values('user_count', ascending=False).reset_index(drop=True)
    return distribution
```

При интерпретации результатов учитывайте размер сегментов: если в сегменте мало пользователей (например, десятки), то доли могут быть нестабильными и недостоверными, то есть показывать широкую вариацию значений.

---


```python
def analyze_user_repeat_rate(df_users, group_col, xlabel, target_col='user_id', shape_is_two_col='is_two', top_n=None):
    
    distribution = df_users.groupby(group_col).agg(
        user_count=(target_col, 'count'),
        share_is_two=(shape_is_two_col, 'mean')
    ).reset_index()
    
    distribution = distribution.sort_values('user_count', ascending=False)
    
    if top_n and len(distribution) > top_n:
        plot_data = distribution.head(top_n)
        print(f"\nПоказаны топ-{top_n} сегментов из {len(distribution)}")
    else:
        plot_data = distribution
    
    plot_data.plot(kind='bar', 
                   x=group_col, 
                   y='share_is_two', 
                   title=f'Доля пользователей с 2+ заказами по {xlabel}',
                   xlabel=xlabel,
                   ylabel='Доля пользователей',
                   figsize=(12, 6),
                   legend=False)
    
    overall_avg = df_users[shape_is_two_col].mean()
    plt.axhline(y=overall_avg, color='red', linestyle='--', linewidth=2, label=f'Среднее: {overall_avg:.2%}')
    plt.legend()
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
    return distribution
```

```python
# Создание сегментов
conditions = [
    (df_users['avg_tickets_count'] >= 1) & (df_users['avg_tickets_count'] < 2),
    (df_users['avg_tickets_count'] >= 2) & (df_users['avg_tickets_count'] < 3),
    (df_users['avg_tickets_count'] >= 3) & (df_users['avg_tickets_count'] < 5),
    (df_users['avg_tickets_count'] >= 5)
]

labels = ['1-2 билета', '2-3 билета', '3-5 билетов', '5+ билетов']

df_users['ticket_segment'] = np.select(conditions, labels, default='Другое')

segment_stats = df_users.groupby('ticket_segment').agg({
    'user_id': 'count', 
    'is_two': ['mean', 'sum']
}).round(3)

segment_stats.columns = ['Всего пользователей', 'Доля повторных', 'Кол-во повторных']
segment_stats['Доля от всех пользователей'] = (segment_stats['Всего пользователей'] / len(df_users) * 100).round(1)

segment_order = ['1-2 билета', '2-3 билета', '3-5 билетов', '5+ билетов']
segment_stats = segment_stats.reindex(segment_order)

print("АНАЛИЗ ПО СЕГМЕНТАМ СРЕДНЕГО КОЛИЧЕСТВА БИЛЕТОВ")
display(segment_stats)
```

```python
features_for_analysis = [
    'first_device_type_canonical',  # тип устройства
    'first_region_name',             # регион
    'first_service_name',            # сервис
    'first_event_type_main',         # тип события
    'avg_revenue_rub',                # средняя выручка
    'avg_tickets_count',               # среднее количество билетов
    'avg_days_between_orders',         # средний интервал между заказами
    'orders_count'                      # число заказов (целевая переменная)
]
df_corr = df_users[features_for_analysis].copy()
df_corr = df_corr.dropna()
# Числовые столбцы (интервальные данные)
interval_cols = ['avg_revenue_rub', 'avg_tickets_count', 'avg_days_between_orders', 'orders_count']

phik_matrix = df_corr.phik_matrix(interval_cols=interval_cols)

# Смотрим корреляцию с orders_count
print("\nКорреляция признаков с числом заказов (orders_count):")
corr_with_orders = phik_matrix['orders_count'].sort_values(ascending=False)
print(corr_with_orders.round(3))
# Создаем тепловую карту
plt.figure(figsize=(10, 8))
# Тепловая карта
sns.heatmap(phik_matrix, 
            annot=True,           # показывать значения
            fmt='.2f',             # формат значений
            cmap='coolwarm',        # цветовая схема
            center=0.5,             # центр цветовой шкалы (т.к. phi_k от 0 до 1)
            square=True,            # квадратные ячейки
            linewidths=0.5,         # линии между ячейками
            cbar_kws={'label': 'Корреляция Phi-k (0-1)'})

plt.title('Корреляционная матрица Phi-k', fontsize=14, pad=20)
plt.tight_layout()
plt.show()
```

```python
features = [
    'first_device_type_canonical',
    'first_region_name',
    'first_service_name',
    'first_event_type_main',
    'avg_revenue_rub',
    'avg_tickets_count',
    'avg_days_between_orders',
    'orders_count'
]

# Числовые признаки
numeric_cols = ['avg_revenue_rub', 'avg_tickets_count', 'avg_days_between_orders', 'orders_count']

# Простая функция для визуализации одного сегмента
def plot_segment_correlation(df_segment, segment_name):
    # Берём только нужные колонки и удаляем пропуски
    df_clean = df_segment[features].dropna()
    # Считаем корреляцию
    corr_matrix = df_clean.phik_matrix(interval_cols=numeric_cols)
    # Рисуем тепловую карту
    plt.figure(figsize=(8, 6))
    sns.heatmap(corr_matrix, 
                annot=True,      # показываем числа
                fmt='.2f',        # формат чисел
                cmap='coolwarm',  # цвета
                center=0.5,       # центр цветовой шкалы
                square=True,      # квадратные клетки
                linewidths=0.5)   # линии между клетками
    
    plt.title(f'{segment_name}\nКорреляция признаков', fontsize=14)
    plt.tight_layout()
    plt.show()

# Создаём сегменты
segment_1 = df_users[df_users['orders_count'] == 1]  # 1 заказ
segment_2_4 = df_users[(df_users['orders_count'] >= 2) & (df_users['orders_count'] <= 4)]  # 2-4 заказа
segment_5_plus = df_users[df_users['orders_count'] >= 5]  # 5+ заказов

# Строим графики для каждого сегмента
plot_segment_correlation(segment_1, "1 заказ")
plot_segment_correlation(segment_2_4, "2-4 заказа")
plot_segment_correlation(segment_5_plus, "5+ заказов")
```
